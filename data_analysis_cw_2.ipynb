```python
X = df.drop('y', axis=1)

y = df['y'].map({'yes':1, 'no':0})
```

```python
df.dtypes
```

## Identify Numerical and Categorical Columns

```python
cat_cols = X.select_dtypes(include='object').columns

num_cols = X.select_dtypes(exclude='object').columns

print("Categorical Columns:")
print(cat_cols)

print("\nNumerical Columns:")
print(num_cols)
```

## One-Hot Encoding of Categorical Variables

```python
X_encoded = pd.get_dummies(X,
                           columns=cat_cols,
                           drop_first=True)
```

## Feature Selection (Remove Duration Variable)

```python
X_encoded = X_encoded.drop('duration', axis=1)
```

```python
# Dataset with duration
X_with_duration = pd.get_dummies(X,
                                 columns=cat_cols,
                                 drop_first=True)

# Dataset without duration
X_without_duration = X_with_duration.drop('duration',
                                          axis=1)
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

## Check Final Dataset Shape

```python
print("Training Shape:", X_train.shape)

print("Testing Shape:", X_test.shape)
```

## Handling Class Imbalance

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)
```

## Import Machine Learning Libraries

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve
```

## Logistic Regression Model

```python
lr_model = LogisticRegression(max_iter=1000)

lr_model.fit(X_train_scaled, y_train)
```

```python
y_pred_lr = lr_model.predict(X_test_scaled)

y_prob_lr = lr_model.predict_proba(X_test_scaled)[:,1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_lr))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_lr))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_prob_lr))
```

## Decision Tree Model

```python
dt_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

dt_model.fit(X_train, y_train)
```

```python
y_pred_dt = dt_model.predict(X_test)

y_prob_dt = dt_model.predict_proba(X_test)[:,1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_dt))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_dt))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_dt))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_prob_dt))
```

## Random Forest Model

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)

y_prob_rf = rf_model.predict_proba(X_test)[:,1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_rf))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_prob_rf))
```

## Support Vector Machine (SVM)

```python
svm_model = SVC(
    #probability=True,
    kernel='linear',
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)
```

```python
y_pred_svm = svm_model.predict(X_test_scaled)

y_score_svm = svm_model.decision_function(X_test_scaled)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_svm))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_svm))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_svm))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_score_svm))
```

## Gradient Boosting Model

```python
gb_model = GradientBoostingClassifier(
    random_state=42
)

gb_model.fit(X_train, y_train)
```

```python
y_pred_gb = gb_model.predict(X_test)

y_prob_gb = gb_model.predict_proba(X_test)[:,1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_gb))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_gb))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_gb))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_prob_gb))
```

## Compare Model Performance

```python
from sklearn.metrics import precision_score, recall_score, f1_score

results = pd.DataFrame({

    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest',
        'SVM',
        'Gradient Boosting'
    ],

    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_svm),
        accuracy_score(y_test, y_pred_gb)
    ],

    'Precision': [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_svm),
        precision_score(y_test, y_pred_gb)
    ],

    'Recall': [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_svm),
        recall_score(y_test, y_pred_gb)
    ],

    'F1-Score': [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_svm),
        f1_score(y_test, y_pred_gb)
    ],

    'ROC-AUC': [
        roc_auc_score(y_test, y_prob_lr),
        roc_auc_score(y_test, y_prob_dt),
        roc_auc_score(y_test, y_prob_rf),
        roc_auc_score(y_test, y_score_svm),
        roc_auc_score(y_test, y_prob_gb)
    ]
})

results
```

## Plot Model Comparison

```python
plt.figure(figsize=(10,5))

ax = sns.barplot(x='Model',
                 y='Accuracy',
                 data=results,
                 palette='viridis')

# Add values on bars
for p in ax.patches:
    ax.annotate(f'{p.get_height():.4f}',
                (p.get_x() + p.get_width()/2., p.get_height()),
                ha='center',
                va='bottom',
                fontsize=10)

plt.title("Model Accuracy Comparison")

plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")

plt.xticks(rotation=15)

plt.show()
```

## ROC Curve Comparison

```python
plt.figure(figsize=(8,6))

# ROC values
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_dt, tpr_dt, _ = roc_curve(y_test, y_prob_dt)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)

# SVM uses decision_function()
fpr_svm, tpr_svm, _ = roc_curve(y_test, y_score_svm)

fpr_gb, tpr_gb, _ = roc_curve(y_test, y_prob_gb)

# Plot ROC curves
plt.plot(fpr_lr, tpr_lr, label='Logistic Regression')
plt.plot(fpr_dt, tpr_dt, label='Decision Tree')
plt.plot(fpr_rf, tpr_rf, label='Random Forest')
plt.plot(fpr_svm, tpr_svm, label='SVM')
plt.plot(fpr_gb, tpr_gb, label='Gradient Boosting')

# Random baseline
plt.plot([0,1], [0,1], 'k--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve Comparison")

plt.legend()

plt.show()
```
